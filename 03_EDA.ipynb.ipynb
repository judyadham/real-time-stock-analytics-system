```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (14,6)
```

```python


BASE_PATH = "/content/drive/MyDrive/Stock_Project/"

cib = pd.read_csv(BASE_PATH + "features_cib.csv")

cae = pd.read_csv(BASE_PATH + "features_cae.csv")

hdb = pd.read_csv(BASE_PATH + "features_hdb.csv")
```

```python
cib["Date"] = pd.to_datetime(cib["Date"])
cae["Date"] = pd.to_datetime(cae["Date"])
hdb["Date"] = pd.to_datetime(hdb["Date"])
```

```python
print("CIB")
display(cib.head())

print("CAE")
display(cae.head())

print("HDB")
display(hdb.head())
```

```python
print("CIB Types")
print(cib.dtypes)

print("\nCAE Types")
print(cae.dtypes)

print("\nHDB Types")
print(hdb.dtypes)
```

```python
print("CIB Missing Values:\n", cib.isnull().sum())
print("\nCAE Missing Values:\n", cae.isnull().sum())
print("\nHDB Missing Values:\n", hdb.isnull().sum())
```

```python
print("CIB Duplicates:", cib.duplicated().sum())
print("CAE Duplicates:", cae.duplicated().sum())
print("HDB Duplicates:", hdb.duplicated().sum())
```

```python
display(cib.describe())
display(cae.describe())
display(hdb.describe())
```

```python
plt.plot(cib["Date"], cib["Close"], label="CIB")
plt.plot(cae["Date"], cae["Close"], label="CAE")
plt.plot(hdb["Date"], hdb["Close"], label="HDB")

plt.title("Closing Price Trend Comparison")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(3,1, figsize=(14,12))

ax[0].plot(cib["Date"], cib["Close"])
ax[0].set_title("CIB Price Trend")

ax[1].plot(cae["Date"], cae["Close"])
ax[1].set_title("CAE Price Trend")

ax[2].plot(hdb["Date"], hdb["Close"])
ax[2].set_title("HDB Price Trend")

plt.tight_layout()
plt.show()
```

```python
plt.plot(cib["Date"], cib["Close"], label="Close")
plt.plot(cib["Date"], cib["MA_7"], label="MA_7")
plt.plot(cib["Date"], cib["MA_30"], label="MA_30")

plt.title("CIB Moving Averages")
plt.legend()
plt.show()
```

```python
plt.plot(cae["Date"], cae["Close"], label="Close")
plt.plot(cae["Date"], cae["MA_7"], label="MA_7")
plt.plot(cae["Date"], cae["MA_30"], label="MA_30")

plt.title("CAE Moving Averages")
plt.legend()
plt.show()
```

```python
plt.plot(hdb["Date"], hdb["Close"], label="Close")
plt.plot(hdb["Date"], hdb["MA_7"], label="MA_7")
plt.plot(hdb["Date"], hdb["MA_30"], label="MA_30")

plt.title("HDB Moving Averages")
plt.legend()
plt.show()
```

```python
risk = {
    "CIB": cib["Return"].std(),
    "CAE": cae["Return"].std(),
    "HDB": hdb["Return"].std()
}

risk_df = pd.DataFrame(risk.items(), columns=["Bank","Volatility"])

sns.barplot(data=risk_df, x="Bank", y="Volatility")
plt.title("Volatility Comparison")
plt.show()
```

```python
most_stable = risk_df.loc[risk_df["Volatility"].idxmin()]
print("Most Stable Bank:", most_stable["Bank"])
print("Risk:", most_stable["Volatility"])
```

```python
profit = {
    "CIB": cib["Return"].mean(),
    "CAE": cae["Return"].mean(),
    "HDB": hdb["Return"].mean()
}

profit_df = pd.DataFrame(profit.items(), columns=["Bank","Avg_Return"])

sns.barplot(data=profit_df, x="Bank", y="Avg_Return")
plt.title("Average Return Comparison")
plt.show()
```

```python
best = profit_df.loc[profit_df["Avg_Return"].idxmax()]
print("Most Profitable Bank:", best["Bank"])
print("Average Return:", best["Avg_Return"])
```

```python
sns.heatmap(cib.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("CIB Correlation Heatmap")
plt.show()
```

```python
Q1 = cib["Return"].quantile(0.25)
Q3 = cib["Return"].quantile(0.75)
IQR = Q3 - Q1

outliers = cib[(cib["Return"] < Q1 - 1.5*IQR) | (cib["Return"] > Q3 + 1.5*IQR)]
print("CIB Outliers:", len(outliers))
```

```python


plt.figure(figsize=(10,6))
sns.heatmap(cae.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("CAE Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(hdb.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("HDB Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Combine ONLY returns
returns_df = pd.DataFrame({
    "CIB": cib["Return"],
    "CAE": cae["Return"],
    "HDB": hdb["Return"]
})

# Correlation between stocks
corr = returns_df.corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Stock Correlation (CIB vs CAE vs HDB)")
plt.show()
```
