```python
import pandas as pd
import numpy as np
```

```python


BASE_PATH = "/content/drive/MyDrive/Stock_Project/"

cib_df = pd.read_csv(BASE_PATH + "features_cib.csv")

cae_df = pd.read_csv(BASE_PATH + "features_cae.csv")

hdb_df = pd.read_csv(BASE_PATH + "features_hdb.csv")
```

```python
datasets = {
    "CIB": cib_df,
    "CAE": cae_df,
    "HDB": hdb_df
}
```

```python
advisor_df = pd.read_csv(BASE_PATH + "investment_advisor_results (1).csv")
```

```python
# =========================
# PHASE 6 — Real-Time Alerts
# =========================

alerts = [] #This list will store all alert messages.

for name, df in datasets.items(): #df = DataFrame containing stock prices

    # =========================
    # Current & Previous Prices
    # =========================

    current_price = df["Close"].iloc[-1] #last row.(today)

    previous_price = df["Close"].iloc[-2] #second last row.(previous day)

    # =========================
    # Price Change %
    # =========================

    price_change = (
        (current_price - previous_price)
        / previous_price
    ) * 100

    # =========================
    # Big Drop Alert
    # =========================

    if price_change < -2:

        alert = (
            f"🚨 BIG DROP ALERT: {name} "
            f"dropped by {price_change:.2f}%"
        )

    # =========================
    # Big Rise Alert
    # =========================

    elif price_change > 2:

        alert = (
            f"📈 BIG RISE ALERT: {name} "
            f"increased by {price_change:.2f}%"
        )

    # =========================
    # Stable Market
    # =========================

    else:

        alert = (
            f"✅ {name} market is stable."
        )

    alerts.append(alert)
```

```python
# =========================
# Show Alerts
# =========================

print("\n")
print("=" * 60)
print("REAL-TIME ALERTS SYSTEM")
print("=" * 60)

for alert in alerts:

    print(alert)
```
