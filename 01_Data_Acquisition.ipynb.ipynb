```python
#CELL 2 — Load Datasets
BASE_PATH = "/content/drive/MyDrive/Stock_Project/"
cib = pd.read_csv(BASE_PATH + "CIB_2010_2026.csv")

cae = pd.read_csv(BASE_PATH + "Credit Agricole Egypt Stock Price History.csv")

hdb = pd.read_csv(BASE_PATH + "Housing_Development_Bank_2010_to_now.csv")
```

```python

#CELL 1 — Import Libraries
import pandas as pd
import numpy as np
```

```python
#CELL 3 — Preview Data
display(cib.head())
display(cae.head())
display(hdb.head())
```

```python
#CELL 4 — Clean Column Names
cib.columns = cib.columns.str.strip()
cae.columns = cae.columns.str.strip()
hdb.columns = hdb.columns.str.strip()

print(cib.columns)
print(cae.columns)
print(hdb.columns)
```

```python
#CELL 5 — Rename Common Columns
rename_map = {
    'Price': 'Close',
    'Vol.': 'Volume',
    'Change %': 'ChangePercent'
}

cib = cib.rename(columns=rename_map)
cae = cae.rename(columns=rename_map)
hdb = hdb.rename(columns=rename_map)
```

```python
#CELL 6 — Remove Fully Empty Rows
cib = cib.dropna(how='all')
cae = cae.dropna(how='all')
hdb = hdb.dropna(how='all')
```

```python
#CELL 7 — Convert Date Column
cib['Date'] = pd.to_datetime(cib['Date'], errors='coerce')
cae['Date'] = pd.to_datetime(cae['Date'], errors='coerce')
hdb['Date'] = pd.to_datetime(hdb['Date'], errors='coerce')
```

```python
#CELL 8 — Remove Invalid Dates
cib = cib.dropna(subset=['Date'])
cae = cae.dropna(subset=['Date'])
hdb = hdb.dropna(subset=['Date'])
```

```python
#CELL 9 — Remove Duplicates
print("CIB duplicates:", cib.duplicated().sum())
print("CAE duplicates:", cae.duplicated().sum())
print("HDB duplicates:", hdb.duplicated().sum())

cib = cib.drop_duplicates()
cae = cae.drop_duplicates()
hdb = hdb.drop_duplicates()
```

```python
#CELL 10 — Convert Numeric Columns
def fix_numeric(df):
    for col in df.columns:
        if col != 'Date':
            df[col] = df[col].astype(str)
            df[col] = df[col].str.replace(',', '', regex=False)
            df[col] = df[col].str.replace('%', '', regex=False)
            df[col] = df[col].str.replace('K', '000', regex=False)
            df[col] = df[col].str.replace('M', '000000', regex=False)
            df[col] = df[col].str.replace('B', '000000000', regex=False)
            df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

cib = fix_numeric(cib)
cae = fix_numeric(cae)
hdb = fix_numeric(hdb)
```

```python
#CELL 11 — Check Missing Values
print("CIB Missing:\n", cib.isnull().sum())
print("\nCAE Missing:\n", cae.isnull().sum())
print("\nHDB Missing:\n", hdb.isnull().sum())
```

```python
#CELL 12 — Fill Missing Values
import numpy as np

def handle_missing(df):
    # Fill numeric columns with median
    numeric_cols = df.select_dtypes(include=np.number).columns

    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].median())

    # Remove rows with missing Date
    if 'Date' in df.columns:
        df = df.dropna(subset=['Date'])

    return df

cib = handle_missing(cib)
cae = handle_missing(cae)
hdb = handle_missing(hdb)
```

```python
#CELL 13 — Fill Remaining Numeric Nulls with Median
def fill_median(df):
    numeric_cols = df.select_dtypes(include=np.number).columns
    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].median())
    return df

cib = fill_median(cib)
cae = fill_median(cae)
hdb = fill_median(hdb)
```

```python
#CELL 14 — Final Missing Check
print("CIB:\n", cib.isnull().sum())
print("\nCAE:\n", cae.isnull().sum())
print("\nHDB:\n", hdb.isnull().sum())
```

```python
#CELL 15 — Sort by Date
cib = cib.sort_values('Date').reset_index(drop=True)
cae = cae.sort_values('Date').reset_index(drop=True)
hdb = hdb.sort_values('Date').reset_index(drop=True)
```

```python
#CELL 16 — Remove Negative Values
for df in [cib, cae, hdb]:
    for col in ['Open', 'High', 'Low', 'Close', 'Volume']:
        if col in df.columns:
            df.drop(df[df[col] < 0].index, inplace=True)
```

```python
#CELL 17 — Final Report
print("CIB Shape:", cib.shape)
print("CAE Shape:", cae.shape)
print("HDB Shape:", hdb.shape)

print("CIB Range:", cib['Date'].min(), "to", cib['Date'].max())
print("CAE Range:", cae['Date'].min(), "to", cae['Date'].max())
print("HDB Range:", hdb['Date'].min(), "to", hdb['Date'].max())
```

```python
#CELL 18 — Save Clean Files
cib.to_csv("clean_cib.csv", index=False)
cae.to_csv("clean_cae.csv", index=False)
hdb.to_csv("clean_hdb.csv", index=False)
```
