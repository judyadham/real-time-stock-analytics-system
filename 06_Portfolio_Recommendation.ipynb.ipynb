```python
import pandas as pd
import numpy as np
```

```python

BASE_PATH = "/content/drive/MyDrive/Stock_Project/"

cib_df = pd.read_csv(BASE_PATH + "features_cib.csv")

cae_df = pd.read_csv(BASE_PATH + "features_cae.csv")

hdb_df = pd.read_csv(BASE_PATH + "features_hdb.csv")
```

```python
advisor_df = pd.read_csv(BASE_PATH+"investment_advisor_results (1).csv")
```

```python
datasets = {

    "CIB": cib_df,

    "CAE": cae_df,

    "HDB": hdb_df
}
```

```python
for name, df in datasets.items(): #It takes the original "Close" prices and increases them randomly between 1% and 10% to simulate future predictions.

    df["Predicted_Close"] = (
        df["Close"] *
        (1 + np.random.uniform(0.01, 0.10, len(df)))
    )

    datasets[name] = df
```

```python
# =========================
# Investment Amount
# =========================

investment_amount = 10000
```

```python
# =========================
# Portfolio Simulator
# =========================

portfolio_results = [] #This list will store the results for all stocks.

for name, df in datasets.items(): #The loop analyzes every stock one by one.

    # Current price
    current_price = df["Close"].iloc[-1] #Gets the latest real stock price.

    # Predicted price
    predicted_price = df["Predicted_Close"].iloc[-1] #Gets the latest predicted future price.

    # Expected return
    expected_return = (
        (predicted_price - current_price)
        / current_price
    )

    # Future value  This estimates how much the investment may become in the future.
    future_value = (
        investment_amount *
        (1 + expected_return)
    )

    # Profit
    profit = (
        future_value -
        investment_amount
    )

    # Save results
    portfolio_results.append({

        "Stock": name,

        "Investment": round(investment_amount, 2),

        "Current Price": round(current_price, 2),

        "Predicted Price": round(predicted_price, 2),

        "Expected Return %": round(expected_return * 100, 2),

        "Future Value": round(future_value, 2),

        "Expected Profit": round(profit, 2)
    })
```

```python
portfolio_df = pd.DataFrame(portfolio_results)

portfolio_df
```

```python
# =========================
# Best Investment
# =========================

best_stock = portfolio_df.loc[
    portfolio_df["Expected Profit"].idxmax()
]

print("\n")
print("=" * 60)

print("🏆 BEST INVESTMENT ALLOCATION")

print("=" * 60)

print(f"📌 Best Stock: {best_stock['Stock']}")

print(f"💰 Expected Profit: {best_stock['Expected Profit']} EGP")

print(f"📈 Future Value: {best_stock['Future Value']} EGP")
```
